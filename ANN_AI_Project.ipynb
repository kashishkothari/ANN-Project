```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sn
```

```python
df = pd.read_csv("/content/Students Annual Score in History Geography Subj.csv")
df.head(2)
```

```python
df.columns
```

```python
df.shape
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Gender']=le.fit_transform(df['Gender'])
```

```python
x = df[['Gender', 'Test_01', 'Exam_01', 'Test_02', 'Exam_02', 'Test_03',
       'Exam_03', 'Annual_Score']]
y = df[['Success']]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,stratify=y)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
x_train = np.asarray(x_train)
y_train = np.asarray(y_train)
x_test = np.asarray(x_test)
y_test = np.asarray(y_test)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten,Activation
from tensorflow.keras.layers import Conv2D, MaxPooling2D,GlobalAveragePooling2D,BatchNormalization
```

```python
classifier=Sequential()
```

```python
# Adding the input layer with 3 hidden  layer
classifier.add(Dense(50, input_shape=(8,), activation='relu')) ## 50 hidden neuron / input_shape is X features
classifier.add(Dropout(0.25))

classifier.add(Dense(20, activation='relu'))
classifier.add(Dropout(0.25))

# Adding the output layer
classifier.add(Dense(1, activation='sigmoid'))  ## output is Binary and 1 column
```

```python
classifier.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model_history=classifier.fit(x_train, y_train, batch_size=32, epochs=50,validation_split=0.33)
```

```python
classifier.predict(x_test)
```

```python
y_predict=classifier.predict(x_test)
```

```python
plt.plot(model_history.history['accuracy'])
plt.plot(model_history.history['val_accuracy'])
plt.title('model accuracy')
plt.ylabel('accuracy')
plt.xlabel('epoch')
plt.legend(['train', 'test'], loc='upper left')
plt.show()
```

```python
plt.plot(model_history.history['loss'])
plt.plot(model_history.history['val_loss'])
plt.title('model loss')
plt.ylabel('loss')
plt.xlabel('epoch')
plt.legend(['train', 'test'], loc='upper left')
plt.show()
```
